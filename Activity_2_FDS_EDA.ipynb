```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
plt.style.use('ggplot')
```

```python
import io
df = pd.read_csv("Practice_HR-Employee-Attrition-All.csv")
df.head()
```

```python
print(df.shape)
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
sns.countplot(x='Attrition', data=df)
plt.title('Employee Attrition Distribution')
plt.show()
print(df['Attrition'].value_counts())
print(df['Attrition'].value_counts(normalize=True))
```

```python
sns.countplot(x='Gender', data=df)
plt.title("Attrition by Gender")
plt.show()
print(df['Gender'].value_counts())
```

```python
sns.countplot(x='Department', data=df)
plt.title("Attrition by Department")
plt.show()
print(df['Department'].value_counts())
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Age'], bins = 30, kde = True)
plt.title("Age Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['MonthlyIncome'], bins = 40, kde = True)
plt.title("Monthly Income Distribution")
plt.show()
```

```python
sns.heatmap(df.isnull(), cbar=False)
plt.show()
```

```python
sns.countplot(x='Gender', hue='Attrition', data=df)
plt.title("Attrition by Gender")
plt.show()
print(pd.crosstab(df['Gender'], df['Attrition']))
```

```python
sns.countplot(x='Department', hue='Attrition', data=df)
plt.title("Attrition by Department")
plt.show()
print(pd.crosstab(df['Department'], df['Attrition']))
```

```python
sns.countplot(x='OverTime', hue='Attrition', data=df)
plt.title("Attrition by OverTime")
plt.show()
print(pd.crosstab(df['OverTime'], df['Attrition']))
```

```python
sns.countplot(x='BusinessTravel', hue='Attrition', data=df)
plt.title("Attrition by Business Travel")
plt.show()
print(pd.crosstab(df['BusinessTravel'], df['Attrition']))
```

```python
sns.boxplot(x=df['Age'])
plt.title("Box Plot of Age")
plt.show
```

```python
sns.boxplot(x=df['MonthlyIncome'])
plt.title("Box Plot of Monthly Income")
plt.show
```

```python
numeric_df=df.select_dtypes(include=np.number)
plt.figure(figsize=(16,10))
sns.heatmap(numeric_df.corr(), annot=True, cmap = 'coolwarm')
plt.title("Correlation Matrix of Key Numeric Features")
plt.show()
```

```python
sns.pairplot(df[['Age','MonthlyIncome','JobLevel','Attrition']], hue='Attrition')
plt.show()
```

```python
sns.boxplot(x='Attrition', y='Age', data = df)
plt.title("Attrition by Age")
plt.show()
```

```python
sns.boxplot(x='Attrition', y='MonthlyIncome', data = df)
plt.title("Attrition by Monthly Income")
plt.show()
```

```python
sns.boxplot(x='Attrition', y='YearsAtCompany', data = df)
plt.title("Attrititon by Years at Company")
plt.show()
```

```python
print(df.groupby('Department')['MonthlyIncome'].mean())
sns.barplot(x='Department', y='MonthlyIncome', data=df)
plt.title("Average Monthly Income by Department")
plt.show()
```

```python
attr_by_role = df.groupby('JobRole')['Attrition'].apply(lambda x: (x=='Yes').mean())
print(attr_by_role.sort_values(ascending = False))
sns.barplot(x=attr_by_role.values, y=attr_by_role.index)
plt.show()
```

```python
bins= [17,25,35,45,55,65]
labels = ['18-25','26-35','36-45','46-55','56-65']
df['AgeGroup'] = pd.cut(df['Age'], bins=bins, labels=labels)
sns.countplot(x='AgeGroup', hue='Attrition', data=df)
plt.title("Attrition by Age Group")
plt.show()
```

```python
df['Income_Group'] = pd.qcut(df['MonthlyIncome'], q=4, labels=['Low', 'Medium', 'High', 'Very High'])
sns.countplot(x='Income_Group', hue='Attrition', data=df)
plt.title("Attrition by Income Group")
plt.show()
```

```python
print(df.groupby('JobLevel')['Age'].mean())
sns.barplot(x='JobLevel',y='Age',data=df)
plt.title("Average Age by Job Level")
plt.show()
```

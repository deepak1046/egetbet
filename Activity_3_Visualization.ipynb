```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("whitegrid")
```

```python
df = pd.read_csv("country_wise_latest.csv")
df.head()
```

```python
plt.figure(figsize=(7,5))
plt.hist(df["Confirmed"], bins=15)
plt.title("Distribution of Confirmed Cases")
plt.xlabel("Confirmed Cases")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.boxplot(y="Deaths / 100 Cases", data=df)
plt.title("Boxplot of Deaths per 100 Cases")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.scatterplot(x="Confirmed", y="Deaths", hue="WHO Region", data=df)
plt.title("Scatterplot of Confirmed Cases vs Deaths")
plt.xlabel("Confirmed Cases")
plt.ylabel("Deaths")
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.barplot(x="WHO Region", y="Deaths / 100 Cases", data=df)
plt.title("Barplot of Deaths per 100 Cases by WHO Region")
plt.xlabel("WHO Region")
plt.ylabel("Deaths per 100 Cases")
plt.xticks(rotation=30, ha="right")
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.countplot(x="WHO Region", data=df, order=df["WHO Region"].value_counts().index)
plt.title("Number of Countries per WHO Regions")
plt.xlabel("WHO Region")
plt.ylabel("Count")
plt.xticks(rotation=30, ha="right")
plt.show()
```

```python
sns.pairplot(df[["Confirmed","Deaths","Recovered","Active","WHO Region"]], hue="WHO Region")
plt.show()
```

```python
plt.figure(figsize=(9,7))
corr = df.select_dtypes(include="number").corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix")
plt.show()
```
